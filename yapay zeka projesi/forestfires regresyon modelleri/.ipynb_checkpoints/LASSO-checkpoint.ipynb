```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Read the CSV file
df = pd.read_csv("./forestfires.csv")  # Ensure the file path is correct
df = df.dropna()

# Create dummy variables for 'month' and 'day'
dms = pd.get_dummies(df[['month', 'day']])

# Define target variable
y = df["area"]

# Drop 'area', 'month', and 'day' from X
X_ = df.drop(['area', 'month', 'day'], axis=1).astype('float')

# Concatenate dummy variables with X
X = pd.concat([X_, dms], axis=1)

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=99)
```

```python
df.shape
```

```python
lasso_model=Lasso().fit(X_train, y_train)
```

```python
lasso_model
```

```python
lasso_model.intercept_
```

```python
lasso_model.coef_
```

```python
#farklı lambda değerlerine karşılık katsayılar oluştur veri setine ekle
```

```python
lasso= Lasso()
coefs= []
alphas=np.random.randint(0,1000,100)
for a in alphas:
    lasso.set_params(alpha=a)
    lasso.fit(X_train,y_train)
    coefs.append(lasso.coef_)
```

```python
#katsayıları görselleştirmek,değerlerin 0'a gidişi
ax=plt.gca()
ax.plot(alphas,coefs)
ax.set_xscale("log")
```

```python
lasso_model
```

```python
#train seti için tahmin
lasso_model.predict(X_train)[0:5]
```

```python
#test seti için bağımlı değişken tahmini
lasso_model.predict(X_test)[0:5]
```

```python
y_pred = lasso_model.predict(X_test)
```

```python
#test hatası
np.sqrt(mean_squared_error(y_test, y_pred))
```

```python
#modelin açıklanabilirliği
r2_score(y_test,y_pred)
```

```python
lasso_cv_model = LassoCV(cv= 10, max_iter=100000).fit(X_train, y_train)
```

```python
lasso_cv_model.alpha_
```

```python
#hata gözlemleme
lasso_tuned = Lasso().set_params(alpha=lasso_cv_model.alpha_).fit(X_train, y_train)
```

```python
y_pred = lasso_tuned.predict(X_test)
```

```python
#final modelinin test hatası
np.sqrt(mean_squared_error(y_test, y_pred))
```

```python
#tune edilmiş verinin katsayıları
pd.Series(lasso_tuned.coef_, index= X_train.columns)
#0 olmayan değerler aranan veriye etki eder
```
