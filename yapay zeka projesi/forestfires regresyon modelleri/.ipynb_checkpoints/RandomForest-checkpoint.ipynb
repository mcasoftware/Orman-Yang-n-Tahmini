```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Read the CSV file
df = pd.read_csv("./forestfires.csv")  # Ensure the file path is correct
df = df.dropna()

# Create dummy variables for 'month' and 'day'
dms = pd.get_dummies(df[['month', 'day']])

# Define target variable
y = df["area"]

# Drop 'area', 'month', and 'day' from X
X_ = df.drop(['area', 'month', 'day'], axis=1).astype('float')

# Concatenate dummy variables with X
X = pd.concat([X_, dms], axis=1)

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=99)
```

```python
rf_model=RandomForestRegressor(random_state=99).fit(X_train, y_train)
```

```python
rf_model
```

```python
#tahmin etme ve test hatası bulma
y_pred =rf_model.predict(X_test)
np.sqrt(mean_squared_error(y_test, y_pred))
```

```python
rf_model=RandomForestRegressor(random_state=99).fit(X_train, y_train)
rf_model
```

```python
rf_params={"max_depth": [5],
          "max_features":[1,2],
          "n_estimators":[200, 500],
          "min_samples_split":[2,10]}
```

```python
rf_cv_model = GridSearchCV(rf_model, rf_params, cv=10, n_jobs= -1 , verbose=2).fit(X_train, y_train)
```

```python
rf_cv_model.best_params_
```

```python
rf_model=RandomForestRegressor(random_state=99,
                              max_depth=5,
                              max_features=1,
                              min_samples_split=2,
                              n_estimators=200)
rf_tuned=rf_model.fit(X_train, y_train)
```

```python
rf_tuned
```

```python
#test hatası
y_pred=rf_tuned.predict(X_test)
np.sqrt(mean_squared_error(y_test,y_pred))
```

```python
#kullandığımız verilerin hangisinin daha önemli olduğunun grafiği
Importance=pd.DataFrame({'Importance':rf_tuned.feature_importances_*100},
                       index=X_train.columns)
Importance.sort_values(by='Importance',
                      axis=0,
                      ascending=True).plot(kind='barh',
                                          color='r',)
plt.xlabel('Variable Importance')
plt.gca().legend_=None
```
